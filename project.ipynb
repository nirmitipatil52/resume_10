```python
from resume_parser import extract_text_from_pdf
from skill_detector import detect_skills
from ats_score import calculate_ats_score

# Extract resume text
text = extract_text_from_pdf("sample_resume.pdf")

# Detect skills
skills = detect_skills(text)

# Select role
target_role = "Data Scientist"

# Calculate ATS score
score, matched, missing = calculate_ats_score(skills, target_role)

print("ATS Score:", score)
print("Matched Skills:", matched)
print("Missing Skills:", missing)
```

```python
from resume_parser import extract_text_from_pdf
from skill_detector import detect_skills
from role_predictor import predict_role

# Extract text
text = extract_text_from_pdf("sample_resume.pdf")

# Detect skills
skills = detect_skills(text)

# Predict role
best_role, best_score, top_roles = predict_role(skills)

print("Best Role:", best_role)
print("Match Score:", best_score)

print("\nTop Matching Roles:")

for role, score in top_roles:
    print(role, ":", score)
```

```python
from suggestion_engine import generate_suggestions

# Generate suggestions
suggestions = generate_suggestions(best_role, missing)

print("Suggestions:")

for s in suggestions:
    print("-", s)
```
